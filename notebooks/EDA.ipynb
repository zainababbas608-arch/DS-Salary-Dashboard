# Exploratory Data Analysis

Visualizing the enriched salary dataset (102K rows, 10 features).

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

os.chdir(r"D:\Desktop\Data Science Salary Dashboard")
sns.set_style("whitegrid")
plt.rcParams["figure.figsize"] = (10, 5)
plt.rcParams["font.size"] = 12
```

```python
df = pd.read_csv("data/processed/salary_model_ready.csv")
print(f"Shape: {df.shape}")
print(f"Columns: {list(df.columns)}")
df.head()
```

```python
df.describe()
```

```python
print("Nulls:")
print(df.isnull().sum())
print(f"\nDuplicates: {df.duplicated().sum()}")
```

## 1. Salary Distribution

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].hist(df["salary_in_usd"], bins=50, edgecolor="black", alpha=0.7, color="steelblue")
axes[0].set_title("Salary Distribution (All)")
axes[0].set_xlabel("Salary (USD)")
axes[0].set_ylabel("Count")
axes[0].axvline(df["salary_in_usd"].median(), color="red", linestyle="--", label=f"Median: ${df['salary_in_usd'].median():,.0f}")
axes[0].legend()

axes[1].hist(df["salary_in_usd"], bins=50, edgecolor="black", alpha=0.7, color="steelblue")
axes[1].set_title("Salary Distribution (Zoomed < $300K)")
axes[1].set_xlabel("Salary (USD)")
axes[1].set_xlim(0, 300000)

plt.tight_layout()
plt.show()
```

## 2. Salary by Experience Level

```python
exp_order = ["Entry", "Mid", "Senior", "Lead", "Executive"]
exp_data = df.groupby("experience_level")["salary_in_usd"].agg(["mean", "median", "count"]).reindex(exp_order)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

exp_data["mean"].plot(kind="bar", ax=axes[0], color="steelblue", edgecolor="black")
axes[0].set_title("Mean Salary by Experience Level")
axes[0].set_ylabel("Mean Salary (USD)")
axes[0].set_xlabel("")
axes[0].tick_params(axis="x", rotation=0)
for i, v in enumerate(exp_data["mean"]):
    axes[0].text(i, v + 1000, f"${v:,.0f}", ha="center", fontsize=10)

df.boxplot(column="salary_in_usd", by="experience_level", ax=axes[1],
           positions=range(len(exp_order)), patch_artist=True)
axes[1].set_title("Salary Spread by Experience Level")
axes[1].set_xlabel("")
axes[1].set_ylabel("Salary (USD)")
axes[1].set_xticklabels(exp_order, rotation=0)
plt.suptitle("")

plt.tight_layout()
plt.show()

print(exp_data)
```

## 3. Salary by Education Level

```python
edu_order = ["High School", "Bachelor's", "Master's", "PhD", "Other"]
edu_data = df.groupby("education_level")["salary_in_usd"].agg(["mean", "median", "count"]).reindex(edu_order)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

edu_data["mean"].plot(kind="bar", ax=axes[0], color="coral", edgecolor="black")
axes[0].set_title("Mean Salary by Education Level")
axes[0].set_ylabel("Mean Salary (USD)")
axes[0].set_xlabel("")
axes[0].tick_params(axis="x", rotation=0)
for i, v in enumerate(edu_data["mean"]):
    axes[0].text(i, v + 1000, f"${v:,.0f}", ha="center", fontsize=10)

edu_data["count"].plot(kind="bar", ax=axes[1], color="coral", edgecolor="black")
axes[1].set_title("Row Count by Education Level")
axes[1].set_ylabel("Count")
axes[1].set_xlabel("")
axes[1].tick_params(axis="x", rotation=0)

plt.tight_layout()
plt.show()

print(edu_data)
```

## 4. Salary by Industry

```python
ind_data = df.groupby("industry")["salary_in_usd"].agg(["mean", "count"]).sort_values("mean", ascending=True)

fig, axes = plt.subplots(1, 2, figsize=(16, 8))

ind_data["mean"].plot(kind="barh", ax=axes[0], color="teal", edgecolor="black")
axes[0].set_title("Mean Salary by Industry")
axes[0].set_xlabel("Mean Salary (USD)")

ind_data["count"].plot(kind="barh", ax=axes[1], color="teal", edgecolor="black")
axes[1].set_title("Row Count by Industry")
axes[1].set_xlabel("Count")

plt.tight_layout()
plt.show()
```

## 5. Salary by Work Setting

```python
ws_order = ["Onsite", "Hybrid", "Remote"]
ws_data = df.groupby("work_setting")["salary_in_usd"].agg(["mean", "median", "count"]).reindex(ws_order)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

ws_data["mean"].plot(kind="bar", ax=axes[0], color="mediumpurple", edgecolor="black")
axes[0].set_title("Mean Salary by Work Setting")
axes[0].set_ylabel("Mean Salary (USD)")
axes[0].set_xlabel("")
axes[0].tick_params(axis="x", rotation=0)
for i, v in enumerate(ws_data["mean"]):
    axes[0].text(i, v + 1000, f"${v:,.0f}", ha="center", fontsize=10)

ws_data["count"].plot(kind="bar", ax=axes[1], color="mediumpurple", edgecolor="black")
axes[1].set_title("Row Count by Work Setting")
axes[1].set_ylabel("Count")
axes[1].set_xlabel("")
axes[1].tick_params(axis="x", rotation=0)

plt.tight_layout()
plt.show()
```

## 6. Salary by Experience Years (Numeric)

```python
fig, ax = plt.subplots(figsize=(12, 5))

exp_bins = pd.cut(df["experience_years"], bins=[0, 2, 5, 10, 15, 20, 30],
                  labels=["0-2", "3-5", "6-10", "11-15", "16-20", "21+"])
exp_salary = df.groupby(exp_bins, observed=True)["salary_in_usd"].agg(["mean", "count"])

ax.bar(exp_salary.index.astype(str), exp_salary["mean"], color="darkorange", edgecolor="black")
ax.set_title("Mean Salary by Years of Experience")
ax.set_xlabel("Years of Experience")
ax.set_ylabel("Mean Salary (USD)")
for i, v in enumerate(exp_salary["mean"]):
    ax.text(i, v + 1000, f"${v:,.0f}", ha="center", fontsize=10)

plt.tight_layout()
plt.show()
```

## 7. Salary by Company Size

```python
size_order = ["Small", "Medium", "Large", "Enterprise"]
size_data = df.groupby("company_size")["salary_in_usd"].agg(["mean", "count"]).reindex(size_order)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

size_data["mean"].plot(kind="bar", ax=axes[0], color="darkgreen", edgecolor="black")
axes[0].set_title("Mean Salary by Company Size")
axes[0].set_ylabel("Mean Salary (USD)")
axes[0].set_xlabel("")
axes[0].tick_params(axis="x", rotation=0)

size_data["count"].plot(kind="bar", ax=axes[1], color="darkgreen", edgecolor="black")
axes[1].set_title("Row Count by Company Size")
axes[1].set_ylabel("Count")
axes[1].set_xlabel("")
axes[1].tick_params(axis="x", rotation=0)

plt.tight_layout()
plt.show()
```

## 8. Salary Trend Over Years

```python
year_data = df.groupby("work_year")["salary_in_usd"].agg(["mean", "median", "count"])

fig, ax1 = plt.subplots(figsize=(10, 5))

ax1.plot(year_data.index, year_data["mean"], "o-", color="steelblue", label="Mean", linewidth=2)
ax1.plot(year_data.index, year_data["median"], "s--", color="coral", label="Median", linewidth=2)
ax1.set_title("Salary Trend Over Years")
ax1.set_xlabel("Year")
ax1.set_ylabel("Salary (USD)")
ax1.legend()
ax1.set_xticks(year_data.index)

ax2 = ax1.twinx()
ax2.bar(year_data.index, year_data["count"], alpha=0.2, color="gray", label="Row Count")
ax2.set_ylabel("Row Count")

plt.tight_layout()
plt.show()
```

## 9. Top 15 Job Titles by Salary

```python
top_jobs = df.groupby("job_title")["salary_in_usd"].agg(["mean", "count"])
top_jobs = top_jobs[top_jobs["count"] >= 100].sort_values("mean", ascending=False).head(15)

fig, ax = plt.subplots(figsize=(10, 7))
top_jobs["mean"].plot(kind="barh", ax=ax, color="steelblue", edgecolor="black")
ax.set_title("Top 15 Job Titles by Mean Salary (min 100 rows)")
ax.set_xlabel("Mean Salary (USD)")
ax.invert_yaxis()
plt.tight_layout()
plt.show()
```

## 10. Top 15 Countries by Salary

```python
top_locs = df.groupby("company_location")["salary_in_usd"].agg(["mean", "count"])
top_locs = top_locs[top_locs["count"] >= 100].sort_values("mean", ascending=False).head(15)

fig, ax = plt.subplots(figsize=(10, 7))
top_locs["mean"].plot(kind="barh", ax=ax, color="darkorange", edgecolor="black")
ax.set_title("Top 15 Countries by Mean Salary (min 100 rows)")
ax.set_xlabel("Mean Salary (USD)")
ax.invert_yaxis()
plt.tight_layout()
plt.show()
```

## 11. Correlation Heatmap

```python
num_cols = ["work_year", "experience_years", "salary_in_usd"]
corr = df[num_cols].corr()

fig, ax = plt.subplots(figsize=(6, 5))
sns.heatmap(corr, annot=True, fmt=".3f", cmap="coolwarm", center=0, ax=ax)
ax.set_title("Correlation Heatmap (Numeric Features)")
plt.tight_layout()
plt.show()
```

## 12. Experience Level x Education Heatmap

```python
pivot = df.pivot_table(values="salary_in_usd", index="experience_level",
                       columns="education_level", aggfunc="mean")
pivot = pivot.reindex(index=["Entry", "Mid", "Senior", "Lead", "Executive"])

fig, ax = plt.subplots(figsize=(10, 5))
sns.heatmap(pivot, annot=True, fmt=".0f", cmap="YlOrRd", ax=ax)
ax.set_title("Mean Salary: Experience Level x Education Level")
plt.tight_layout()
plt.show()
```

## 13. Work Setting x Experience Level

```python
pivot2 = df.pivot_table(values="salary_in_usd", index="experience_level",
                        columns="work_setting", aggfunc="mean")
pivot2 = pivot2.reindex(index=["Entry", "Mid", "Senior", "Lead", "Executive"])

fig, ax = plt.subplots(figsize=(8, 5))
sns.heatmap(pivot2, annot=True, fmt=".0f", cmap="YlGnBu", ax=ax)
ax.set_title("Mean Salary: Experience Level x Work Setting")
plt.tight_layout()
plt.show()
```

## 14. Industry x Education Heatmap

```python
pivot3 = df.pivot_table(values="salary_in_usd", index="industry",
                        columns="education_level", aggfunc="mean")

fig, ax = plt.subplots(figsize=(10, 8))
sns.heatmap(pivot3, annot=True, fmt=".0f", cmap="RdYlGn", ax=ax)
ax.set_title("Mean Salary: Industry x Education Level")
plt.tight_layout()
plt.show()
```

## 15. Experience Years vs Salary Scatter

```python
fig, ax = plt.subplots(figsize=(10, 6))
sample = df.sample(n=min(5000, len(df)), random_state=42)
scatter = ax.scatter(sample["experience_years"], sample["salary_in_usd"],
                    c=sample["experience_years"], cmap="viridis", alpha=0.3, s=10)
ax.set_title("Experience Years vs Salary")
ax.set_xlabel("Years of Experience")
ax.set_ylabel("Salary (USD)")
plt.colorbar(scatter, label="Experience Years")
plt.tight_layout()
plt.show()
```
