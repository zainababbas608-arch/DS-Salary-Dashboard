```python
exp_level = df["experience_level"].unique()
emp_type = df["employment_type"].unique()
com_size = df["company_size"].unique()

print("Experience levels =", exp_level)
print("Employment Types =", emp_type)
print("Company Sizes = ", com_size)
```

```python
df["experience_level"] = df["experience_level"].replace('EN', "Entry Level")
df["experience_level"] = df["experience_level"].replace('EX', "Expert Level")
df["experience_level"] = df["experience_level"].replace('MI', "Mid Level")
df["experience_level"] = df["experience_level"].replace('SE', "Senior Level")

# df["experience_level"]

df["employment_type"] = df["employment_type"].replace('FT', "Full Time")
df["employment_type"] = df["employment_type"].replace('PT', "Part Time")
df["employment_type"] = df["employment_type"].replace('CT', "Contract")
df["employment_type"] = df["employment_type"].replace('FL', "Freelance")

# df["employment_type"]

df["company_size"] = df["company_size"].replace('L', "Large")
df["company_size"] = df["company_size"].replace('M', "Mid")
df["company_size"] = df["company_size"].replace('S', "Small")

df["company_size"]
```

```python
avg_sal_per_year = df.groupby("work_year")["salary_in_usd"].mean().reset_index()
avg_sal_per_year
```

```python
plt.plot(avg_sal_per_year["work_year"], avg_sal_per_year["salary_in_usd"], marker = "o", color = "teal")
plt.xticks(avg_sal_per_year["work_year"])
plt.grid(True)
plt.xlabel("Change in Year")
plt.ylabel("Average Salary (USD)")
plt.title("Average Salary per Year")
```

```python
avg_sal_per_exp = df.groupby("experience_level")["salary_in_usd"].mean().reset_index()
avg_sal_per_exp
```

```python
plt.plot(avg_sal_per_exp["experience_level"], avg_sal_per_year["salary_in_usd"], marker = "o", color = "blue")
plt.xticks(avg_sal_per_exp["experience_level"])
plt.grid(True)
plt.xlabel("Change in Experience")
plt.ylabel("Average Salary (USD)")
plt.title("Average Salary by Experience Level")
```

```python
avg_sal_per_emptype = df.groupby("employment_type")["salary_in_usd"].mean().reset_index()
avg_sal_per_emptype
```

```python
plt.plot(avg_sal_per_emptype["employment_type"], avg_sal_per_emptype["salary_in_usd"], marker = "o", color = "orange")
plt.xticks(avg_sal_per_emptype["employment_type"])
plt.grid(True)
plt.xlabel("Employment Type")
plt.ylabel("Average Salary (USD)")
plt.title("Average Salary by Type of Employment")
```

```python
avg_sal_per_comsize = df.groupby("company_size")["salary_in_usd"].mean().reset_index()
avg_sal_per_comsize
```

```python
plt.plot(avg_sal_per_comsize["company_size"], avg_sal_per_comsize["salary_in_usd"], marker = "o", color = "purple")
plt.xticks(avg_sal_per_comsize["company_size"])
plt.grid(True)
plt.xlabel("Company Size")
plt.ylabel("Average Salary (USD)")
plt.title("Average Salary by Company Size")
```

```python
avg_sal_emptype_year = df.groupby(["employment_type", "work_year"])["salary_in_usd"].mean().reset_index().sort_values(by= "salary_in_usd", ascending=False)
avg_sal_emptype_year
```

```python
plt.bar(range(len(avg_sal_emptype_year)), avg_sal_emptype_year["salary_in_usd"], color = "darkcyan")
plt.xticks(range(len(avg_sal_emptype_year)), avg_sal_emptype_year["employment_type"] + "-" + avg_sal_emptype_year["work_year"].apply(str), rotation = 45, ha= "right")
plt.title("Average Salary by Employment Type per Year")
plt.xlabel("Employment Type-Year")
plt.ylabel("Average Salary(USD)")
```

```python
avg_sal_exp_year = df.groupby(["experience_level", "work_year"])["salary_in_usd"].mean().reset_index().sort_values(by= "salary_in_usd", ascending=False)
avg_sal_exp_year
```

```python
plt.bar(range(len(avg_sal_exp_year)), avg_sal_exp_year["salary_in_usd"], color = "green")
plt.xticks(range(len(avg_sal_exp_year)), avg_sal_exp_year["experience_level"] + "-" + avg_sal_exp_year["work_year"].apply(str), rotation = 45, ha= "right")
plt.title("Average Salary by Experience Level per Year")
plt.xlabel("Experience Level-Year")
plt.ylabel("Average Salary(USD)")
```
